```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib as mpl
```

```python
df = pd.read_csv("/gpfs/data/shenhavlab/users/ca3261/Visionary_AI/src/analysis/final_figures/BL_exclusion.csv")
```

```python
df = df.rename(columns={"Unnamed: 0": "Stage"})
df = df.set_index("Stage")
plot_df = df.T.reset_index().rename(columns={"index": "Population"})
plot_long = plot_df.melt(
    id_vars="Population",
    var_name="Stage",
    value_name="Count"
)
```

```python
dpi = 300
font_size = 12
fig_size = 5
mpl.rcParams['pdf.fonttype'] = 42
mpl.rcParams['ps.fonttype'] = 42
mpl.rcParams['svg.fonttype'] = 'none'
plt.rcParams["font.family"] = "DejaVu Sans"
```

```python
palette = sns.color_palette("rocket_r", 8)
plt.figure(figsize=(10, 3))
ax = sns.barplot(
    data=plot_long,
    x="Population",
    y="Count",
    hue="Stage",
    palette=palette,
)

for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3, fontsize=10)

plt.title("Stable-Optimized Model BL Exclusion")
plt.xlabel("")
plt.ylabel("Count")
plt.ylim(0, 110)
plt.legend(frameon=False)
ax.get_legend().remove() 
plt.savefig('/gpfs/data/shenhavlab/users/ca3261/Visionary_AI/src/analysis/final_figures/final_submission/extended_data/fig_2/b.pdf') 

plt.show()
```
